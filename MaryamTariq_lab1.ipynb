```python

flight_ids = [101, 102, 103, 104]
departures = ["Karachi", "Lahore", "Islamabad", "Karachi"]
destinations = ["Lahore", "Karachi", "Karachi", "Islamabad"]
prices = [25000, 25500, 30000, 35000]

print("Available Flights: \n1-Karachi to Lahore \n2-Lahore to Karachi \n3-Islamabad to Karachi \n4-Karachi to Islamabad")
departure_city = input("\nEnter departure city: ")

destination_city = input("Enter destination city: ")

found_count = 0

for i in range(4):
    if departures[i] == departure_city and destinations[i] == destination_city:
        found_count = found_count + 1

if found_count == 0:
    print("No flights available")
    exit()

print("\nAvailable Flights:")
for i in range(4):
    if departures[i] == departure_city and destinations[i] == destination_city:
        print("Flight ID:", flight_ids[i], "| Price: Rs.", prices[i])

chosen_id = int(input("\nEnter Flight ID to book: "))

is_booked = False
for i in range(4):
    if flight_ids[i] == chosen_id and departures[i] == departure_city and destinations[i] == destination_city:
        is_booked = True
        break

if is_booked:
    print("Flight booked successfully")
else:
    print("Invalid Flight ID selected.")
```
